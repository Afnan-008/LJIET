```python
def fun(n):
    print("Hello!",n)
n = str(input("Enter Name"))
fun(n)
```

```python
def sum(a,b):
    return a+b
print("Addition of 10 + 20 is",sum(10,20))
```

```python
#Global Variable
a = 0
def sub(a):
    print("Subtraction : ",a-2)
sub(a=10)
```

```python
#Local Variable
def local():
    x = 21
print(x) # It will generate error as x is a local variable and we are accessing it outside the function.
```

```python
# Function with No parameter and with return value
def name():
    name = "Afnan"
    return name
name()
```

```python
# Function with parameter and with return value
def sub(a,b):
    s=a-b
    return s
print("Sub:",sub(10,5))
```

```python
# Function with No parameter and No return value
def printline():
    s=input("Enter name:")
    print(s)
printline()
```

```python
# Function with parameter and No return value
def sum(a,b):
    print("Sum :",a+b)
sum(10,20)
```

```python
def add(a,b):
    a+b
print(add(10,20))
```

```python
def sqr(x):
    """Argument Passed in x, returns square of x"""
    return x*x
n = int(input("Enter Number"))
print(sqr(n))
print(sqr.__doc__)
```

```python
def sum(*n):
    total=0
    for n1 in n:
        total=total+n1
    print("Total",total)
sum(10)
sum(10,20,30)
```

```python
def f1(n1,*s):
    print(n1)
    for s1 in s:
        print(s1)
f1(10)
f1(10,20,30)
```

```python
def f1(*s,n1):
    print(n1)
    for s1 in s:
        print(s1)
f1(n1=10)
f1(10,20,n1=30)
```

```python
def display(**n):
    for k,v in n.items():
        print(k,"=",v)
display(p1="Max verstappen",p2="Lando norris",p3="Oscar piastri")
```

```python
def outer():
    print("Outer Function")
    def inner():
        print("Inner Function")
    inner()
outer()
```
