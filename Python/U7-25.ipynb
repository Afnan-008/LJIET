## Deleting the Directory or Files.
- OS Module provides different methods of removing directories and files:
  - **os.remove()** : this method is used to remove or delete a file, this method cannot delete a directory. if specified path is a directory then **OSError** will be raised.
  - **os.rmdir()** : this metod is used to delete a directory only if it is empty else it raises error.


```python
import os
os.chdir("D:\\Afnan\\B9\\PY")
print(os.listdir())

os.remove("hello.txt")
os.listdir()
```

# ✨Modules
- A module is a file containing python defination and statements. a module can define function, class and variables. it makes the code logically organised.

```python
%%writefile mymod.py
def add(a,b):
    print(a+b)
def sub(a,b):
    print(a-b)
def mul(a,b):
    print(a*b)
def div(a,b):
    print(a/b)
def power(a,b):
    print(a**b)
def floor(a,b):
    print(a//b)
def modulo(a,b):
    print(a%b)
```

```python
import math
print(math.sqrt(9))
```

### ✨ datetime module
- date represents a calender date(year,month,day).
- time represents a time of the day(hour,mins,seconds and mili-seconds) and optional timezone.
- datetime combines both date and time in formation.

```python
from datetime import date
today = date.today()
print(today)

year,month,day = today.year,today.month,today.day
print(year,month,day)

d = date(2026,10,2)
print(d)
```

```python
from datetime import date
today = date.today()
print(today.strftime("%d %b %Y (%A)"))
print(today.strftime("%D %B %y %a"))

print(today.strftime("%d/%B/%y %a"))
print(today.strftime("%d/%b/%Y %A"))

print(today.strftime("%d/%m/%y"))
print(today.strftime("%d/%m/%Y"))
```

```python
from datetime import time
t = time(12,10,0)
print(t)
print(type(t))
```

```python
import time
from datetime import datetime

try:
    while True:
        # Get the current date and time
        now = datetime.now()

        # Format the time as HH:MM:SS (e.g., 09:03:45)
        current_time = now.strftime("%H:%M:%S")

        # Print on the same line using '\r' and flush the output buffer
        print(f"\rCurrent Time: {current_time}", end="", flush=True)

        # Wait for 1 second before updating
        time.sleep(1)
except KeyboardInterrupt:
    print("\nClock stopped.")
```
