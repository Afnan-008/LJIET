# 1.Arithmatic oprater

```python
var1=3
var2=2
print(var1+var2,'add')
print(var1-var2,'sub')
print(var1*var2,'mult')
print(var1/var2,'div')
print(var1%var2,'modulo')
print(var1//var2,'floor')
print(var1**var2,'power')
```

# 2.Assignment oprater

```python
x=5
x+=5
print(x)

y=5
y-=5
print(y)

x=3
x/=3
print(x)

y=2
y//=3
print(y)

y=2
y**=3
print(y)

```

# 3.Relational oprater

```python
var1=5
var2=6
print(var1==var2)
print(var1!=var2)
print(var1>var2)
print(var1<var2)
print(var1>=var2)
print(var1<=var2)
print(1=='1.0')
```
