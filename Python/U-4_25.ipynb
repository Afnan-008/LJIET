```python
s = "P@##@A FRSWAF asd#$ %$# 4325 623"
alphabets = 0
upper = 0
lower = 0
digit = 0
special = 0
space=0
for i in s:
    if i.isalpha():
        alphabets+=1
        if i.isupper():
            upper+=1
        if i.islower():
            lower+=1
    elif i.isdigit():
        digit+=1
    elif i.isspace():
        space+=1
    else:
        special+=1
print("Alphabets :",alphabets)
print("Uppercase Alphabets :",upper)
print("Lowercase Alphabets :",lower)
print("Digits :",digit)
print("Special Symbols :",special)
print("Spaces :",space)
```

```python
# WAP to check whether the password is valid or not.
import string
def password():
    s = input("Enter Password : ")
    cap = False
    special = False
    num = False
    lower = False
    ex = True
    l = len(s)
    if l >= 8:
        for i in s:
            if i.isupper():
                cap = True
            if i.islower():
                lower = True
            if i.isdigit():
                num = True
            if i in string.punctuation:
                if i == "_" or i == "$" or i == "@":
                    ex = True
                else:
                    ex = False
                    break
                
                
    if cap == True and num == True and lower == True and ex == True:
        print("Password is valid")
    else:
        print("Invalid Password!")
password()         
```

```python
s = input("Enter Email :")
if "@" in s:
    t = s.split("@")
    uname = t[0]
    domain = t[1]
    fname = ""
    l = len(uname)
    if l > 1:
        for i in range(0,l):
            if i == 0:
                fname += uname[i]
            else:
                fname += "*"
        print(fname+"@"+domain)    
    else:
        print(s)
else:
    print("email not valid!")
```
