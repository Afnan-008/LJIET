```python
def iscap(n):
    s=n[0]
    t=n[1:]
    if s.isupper() and t.islower():
        return True
    else:
        return False
print(iscap("Afnan Abcs Aytc"))
```

```python
s = "P@##@A FRSWAF asd#$ %$# 4325 623"
alphabets = 0
upper = 0
lower = 0
digit = 0
special = 0
space=0
for i in s:
    if i.isalpha():
        alphabets+=1
        if i.isupper():
            upper+=1
        if i.islower():
            lower+=1
    elif i.isdigit():
        digit+=1
    elif i.isspace():
        space+=1
    else:
        special+=1
print("Alphabets :",alphabets)
print("Uppercase Alphabets :",upper)
print("Lowercase Alphabets :",lower)
print("Digits :",digit)
print("Special Symbols :",special)
print("Spaces :",space)
```
