## OOP & Exception Handeling

#### Exception Handeling

Error in python can be of 2 types:
1) Compile Time(syntax error, indentation error etc...)
2) Run Time

- Errors are problems in a program due to which the program will stop execution.
- on the other hand, exceptions are raised  

### try, except, else & finally
- The try block lets you execute a block of code that can generate errors.
- The except block lets you handle errors.
- The else block lets you execute code when there is no error.
- The finally block lets you execute the code regardless of try and except block.

```python
def div(a,b):
    try:
        a/b
    except:
        print("Cannot Divide by 0.")
    else:
        print(a/b) 
    finally:
        print("This always prints!")

div(5,2)
print()
div(10,8)
print()
div(10,9999)
print()
div(10,0)
```

# Raise an Exception
- To Create Error Manually

```python
def add(a,b):
    if type(a)==str:
        raise TypeError(a,"is not integer")
    if type(b)==str:
        raise TypeError(b,"is not integer")
    return a+b
print(add(10,20))
print(add('50','100'))
```

# Custom Exception
- To generate custom errors

```python
class afnan(Exception):
    def __init__(self,arg):
        self.msg = arg
```

```python
def pri(a):
    for i in a:
        if i.lower()=='afnan':
            raise afnan("Don't Use My Name")
        print(i)

li = ["Mufaddal","Vruj","Pranav","Afnan"]
print(pri(li))
```

```python
def checkbalance():
    money = int(input("Enter Balance:"))
    withdraw = int(input("Enter Amount to Withdraw :"))
    try:
        balance = money-withdraw
        if balance <= 2000:
            raise afnan("Low Balance Alert!!!!")
        print(balance)
    except afnan as e:
        print(e)
checkbalance()
```

### OOP

#### Class
- User Defined Data Structure that binds data members and methods into a single unit.
- Class is a blueprint or code template for object creation.
- Using a class you can create as many objects as you want.

#### Object
- It is an instance of a class.
- It is collection of attributes and methods.
- We use the objects of a class to perform operations.
