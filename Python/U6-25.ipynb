# Working with files

#### Two types of files:
###### Text file and Binary files(not in syllabus)

### File Access Modes :
##### Read only: r
##### Write only: w
##### Read and Write: r+ "Preserves existing data."
##### Write and Read: w+ "Erases (truncates) all existing data immediately upon opening."
##### Append only: a
##### Append and Read: a+

```python
file = open("simple.txt","r")
print(file)
print(file.read())
print(file.closed)

file.close()
print(file.closed)
```

```python
# closes file automatically
with open("simple.txt","r") as f:
    print(f.read())
    print(f.closed)
print(f.closed)
```

```python
# here in 1st time the cursor moves to the end of data so in 2nd time nothing appears as cursor or pointer is at the end.
file = open("simple.txt","r")
print(file.tell())
print(file.read())
print(file.tell())
print(file.read())
file.close()
```

```python
# using seek we can move the cursor back.
file = open("simple.txt","r")
print(file.tell())
print(file.read())

file.seek(10)
print(file.tell())
print(file.read())
file.close()
```

```python
while True:
    print("\n 1. Write")
    print("\n 2. Read")
    print("\n 3. Append")
    print("\n 4. Exit")
    ch = int(input("Enter Choice :"))

    if ch==1:
        with open("data.txt","w") as f:
            data = input("Enter Data :")
            f.write(data+"\n")
            l = eval(input("Enter List of Strings :"))
            f.writelines(l)
        print("---Data Written Success---")

    elif ch==2:
        with open("data.txt","r") as f:
            print(f.read())
            print(f.readline())
            print(f.readlines())
        print("---Data Read Success---")
    elif ch==3:
        with open("data.txt","a") as f:
            data = input("Enter data for append :")
            f.write("\n"+data)
        print("---Data Append Success---")
    elif ch==4:
        print("---Program Ended---")
        break
    else:
        print("---Invalid Choice---")
```

```python
# WAP to copy the content of a file in another file performing line wise reverse
f = open("simple.txt","r")
l = f.readlines()
f.close()
with open("new.txt","w+") as t:
    for i in range(len(l)-1,-1,-1):
        if "\n" not in l[i]:
            t.write(l[i]+"\n")
        else:
            t.write(l[i])
    t.seek(0)
    print(t.read())
```

```python
# WAP to replace a word in the given text file.
t = []
with open("1234.txt","r") as f:
    l = f.readlines()
    for i in l:
        if "python" in i:
            i = i.replace("python","java")
        t.append(i)
with open("1234.txt","r+") as f:
    f.writelines(t)
    f.seek(0)
    print(f.read())
```

```python
# WAP to replace a word in the given text file.
f = open("1234.txt")
x = f.read()
print(x)
if "java" in x:
    x = x.replace("java","python")
elif "python" in x:
    x = x.replace("python","java")
f.close()

f = open("1234.txt","w")
f.write(x)
print()
print(x)
f.close()
```

```python
# WAP to count number of lines, words, characters and spaces from given test file.
f = open("1234.txt")
l = f.readlines()
f.seek(0)
w = f.read()
ch = 0
space = 0
for i in w:
    if (i.isalpha() or i.isdigit()) and i!="\n":
        ch+=1
    elif i.isspace():
        space+=1
    
words = []
word = w.split()
for i in word:
    if not i.isdigit():
        words.append(i)

print("No of lines :",len(l))
print("Character Count :",ch)
print("Space Count :",space)
print("Words Count :",len(words))
```

```python
f = open("new.txt","r")
lines = f.readlines()
f.close()
code = []
for i in lines:
    word=""
    for j in i:
        if j == "#":
            break
        else:
            word+=j
    code += "\n"+word
t = ""
for i in code:
    t+=i

ot = open("output.txt","w+")
ot.write(t)
ot.seek(0)
print(ot.read())
ot.close()
```
