# Working with files

#### Two types of files:
###### Text file and Binary files(not in syllabus)

### File Access Modes :
##### Read only: r
##### Write only: w
##### Read and Write: r+ "Preserves existing data."
##### Write and Read: w+ "Erases (truncates) all existing data immediately upon opening."
##### Append only: a
##### Append and Read: a+

```python
while True:
    print("\n 1. Write")
    print("\n 2. Read")
    print("\n 3. Append")
    print("\n 4. Exit")
    ch = int(input("Enter Choice :"))

    if ch==1:
        with open("data.txt","w") as f:
            data = input("Enter Data :")
            f.write(data+"\n")
            l = eval(input("Enter List of Strings :"))
            f.writelines(l)
        print("---Data Written Success---")

    elif ch==2:
        with open("data.txt","r") as f:
            print(f.read())
            print(f.readline())
            print(f.readlines())
        print("---Data Read Success---")
    elif ch==3:
        with open("data.txt","a") as f:
            data = input("Enter data for append :")
            f.write("\n"+data)
        print("---Data Append Success---")
    elif ch==4:
        print("---Program Ended---")
        break
    else:
        print("---Invalid Choice---")
```

```python
# WAP to copy the content of a file in another file performing line wise reverse
f = open("simple.txt","r")
l = f.readlines()
f.close()
with open("new.txt","w+") as t:
    for i in range(len(l)-1,-1,-1):
        if "\n" not in l[i]:
            t.write(l[i]+"\n")
        else:
            t.write(l[i])
    t.seek(0)
    print(t.read())
```

```python
# WAP to count number of lines, words, characters and spaces from given test file.
f = open("1234.txt")
l = f.readlines()
f.seek(0)
w = f.read()
ch = 0
space = 0
for i in w:
    if (i.isalpha() or i.isdigit()) and i!="\n":
        ch+=1
    elif i.isspace():
        space+=1
    
words = []
word = w.split()
for i in word:
    if not i.isdigit():
        words.append(i)

print("No of lines :",len(l))
print("Character Count :",ch)
print("Space Count :",space)
print("Words Count :",len(words))
```

```python
# WAP to read line by line from given files: file 1 and file 2 and write into file 3.
f1 = open("f1.txt","r")
f2 = open("f2.txt","r")
f3 = open("f3.txt","w+")
d1=[]
d2=[]
for i in f1:
    if "\n" not in i:
        i+="\n"
    d1.append(i)
for i in f2:
    if "\n" not in i:
        i+="\n"
    d2.append(i)
l = len(d1) if len(d1)<len(d2) else len(d2)
for i in range(l):
    data1 = d1[i]
    data2 = d2[i]
    f3.write(data1)
    f3.write(data2)
    t = i
if len(d1) > len(d2):
    for i in range(t,len(d1)):
        data = d1[i]
        f3.write(data)
elif len(d2) > len(d1):
    for i in range(t+1,len(d2)):
        data = d2[i]
        f3.write(data)
f3.seek(0)
print(f3.read())
f1.close()
f2.close()
f3.close()
```

```python
# WAP to reverse the content of one file and store it in second file and also convert content of second file in to upper case
# and store it in third file. also count no of vowels in third file and also print only second line from the content of third file.

f1 = open("para.txt")
data1 = f1.read()
f1.seek(0)
print("\nFile 1 Data :\n",data1)

f2 = open("f-2.txt","w+")
f3 = open("f-3.txt","w+")
rev = data1[::-1]
f2.write(rev)
f2.seek(0)
print("\nFile 2 Data :\n",f2.read())
f2.seek(0)

f3.write(rev.upper())
vc = 0
f3.seek(0)
print("\nFile 3 Data :\n",f3.read())
f3.seek(0)
fl=""
for i in rev.upper():
    if i != "\n":
        fl+=i
    if i in "AEIOU":
        vc+=1
print("\nNo of vowels in File 3 :",vc)
l = f3.readlines()
print("Second line of file 3:",l[1])
f1.close()
f2.close()
f3.close()
```

```python
#WAP a program to read a text file and print content of text file with word strting with 'i' in reverse order.
f = open("para1.txt")
data = f.read().split()
new = ""
org = ""
for i in data:
    org+=i+" "
    if i.startswith("i") or i.startswith("I"):
        new += i[::-1]+" "
    else:
        new+=i+" "
print(org,"\n")
print(new)
f.close()
```

```python
with open("students.txt","a+") as f:
    while True:
        print("1 to Add Student \n Enter 2 to Display data \n 3 to Exit.")
        ch = int(input("Enter Choice :"))
        if ch==1:
            roll = int(input("Enter roll no :"))
            name = input("Enter Name")
            marks = int(input("Enter Marks :"))
            m = str(roll) + "," + name + "," + str(marks) + "\n"
            f.write(m)
        elif ch==2:
            f.seek(0)
            li = f.readlines()
            for i in li:
                t = i.split(",")
                print("|",t[0],"| ",t[1],"|",t[2])
        elif ch==3:
            break
        else:
            print("Enter Valid Choice!")
```

```python
with open("emp.txt","a+") as f:
    while True:
        print("1 to Add Employee \n Enter 2 to Display data \n 3 to Exit.")
        ch = int(input("Enter Choice :"))
        if ch==1:
            age = int(input("Enter Age :"))
            name = input("Enter Name :")
            salary = int(input("Enter Salary :"))
            m = str(age) + "," + name + "," + str(salary) + "\n"
            f.write(m)
        elif ch==2:
            f.seek(0)
            li = f.readlines()
            avg=[]
            for i in li:
                t = i.split(",")
                print("|",t[0],"| ",t[1],"|",t[2])
                avg.append(int(t[2]))
            average = sum(avg)/len(avg)
            print("Average Salary :",average)
        elif ch==3:
            break
        else:
            print("Enter Valid Choice!")
```

```python
with open("mbox-short.txt","r") as f:
    emails = []
    for i in f.readlines():
        if i.startswith("From:"):
            emails.append(i.split()[1])
    print("All emails:",emails)

    em,c = "",0
    for i in emails:
        if emails.count(i) > c:
            c = emails.count(i)
            em = i
    print("\n",em,c)
```

```python
with open("mbox-short.txt","r") as f:
    senders = []
    receivers = []
    for i in f.readlines():
        if i.startswith("From:"):
            t = i.split()
            senders.append(t[1])
        elif i.startswith("To:"):
            receivers.append(i.split()[1])
    for j in range(len(senders)):
       print(senders[j],"-->",receivers[j])
```
