# Working with files

#### Two types of files:
###### Text file and Binary files(not in syllabus)

### File Access Modes :
##### Read only: r
##### Write only: w
##### Read and Write: r+ "Preserves existing data."
##### Write and Read: w+ "Erases (truncates) all existing data immediately upon opening."
##### Append only: a
##### Append and Read: a+

```python
file = open("simple.txt","r")
print(file)
print(file.read())
print(file.closed)

file.close()
print(file.closed)
```

```python
# closes file automatically
with open("simple.txt","r") as f:
    print(f.read())
    print(f.closed)
print(f.closed)
```

```python
# here in 1st time the cursor moves to the end of data so in 2nd time nothing appears as cursor or pointer is at the end.
file = open("simple.txt","r")
print(file.tell())
print(file.read())
print(file.tell())
print(file.read())
file.close()
```

```python
# using seek we can move the cursor back.
file = open("simple.txt","r")
print(file.tell())
print(file.read())

file.seek(10)
print(file.tell())
print(file.read())
file.close()
```

```python
f = open("utf8_demo.txt","r",encoding='utf-8')
print(f.read())
f.close()
```

```python
f = open("simple.txt","r+")
print(f.read())

print()
f.write("\nHello this is new text added.")
f.seek(0)
print(f.read())
f.close()
```

```python
f = open("simple.txt","r+",encoding='utf-8')
f.write("This is demo file")
print(f.read())
f.close()
```

```python
f = open("abc.txt","w",encoding='utf-8') # creates new file as abc.txt doesn't exist.
f.write("This is demo file\n")
f.write("This is test data\n")
f.write("This is test data 1")
f.close()
```

```python
f = open("abc.txt","w+",encoding='utf-8') # creates new file as abc.txt doesn't exist.
f.write("This is demo file\n")
f.write("This is test data\n")
f.write("This is test data 1")
f.seek(0)
print(f.read())
f.close()
```

```python
f = open("txt.txt","w+",encoding='utf-8') # creates new file as abc.txt doesn't exist.
f.write("This is demo file\n")
f.write("This is test data\n")
f.write("This is test data 1")
f.close()
```

```python
f = open("txt1.txt","a+",encoding='utf-8') # creates new file as abc.txt doesn't exist.
f.write("This is demo file\n")
f.write("This is test data\n")
f.write("This is test data 1\n")
f.seek(0)
f.write("This is test data 2\n") # does not add data at the beginning even after seek(0)

f.seek(0)
print(f.read())
f.close()
```

* File Functions:
    1) **read()** : returns read bytes in the form of string.
    2) **readline()** : reads the line in the file and returns in the form of string.
    3) **readlines()** : reads all the lines in the file and returns in the form of string element in a list.

```python
f = open("simple.txt")
print(f.read())
print()
f.seek(0)
print(f.read())
f.close()
```

```python
f = open("simple.txt")
print(f.read(4))
print(f.read(3))
print(f.read(7))
print(f.read())
f.close()
```
