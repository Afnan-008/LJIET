```python
year = int(input("Enter Year :"))
y = {1:"Jan",2:"Feb",3:"Mar",4:"Apr",5:"May",6:"Jun",7:"Jul",8:"Aug",9:"Sep",10:"Oct",11:"Nov",12:"Dec"}
leap_year=False
if year%4==0 and (year%400==0 or year%100!=0):
    leap_year=True

month = int(input("Enter Month(1-12) :"))
if month in (1,3,5,7,8,10,12):
    length = 31
elif month == 2:
    if leap_year:
        length = 29
    else:
        length = 28
elif month in (2,4,6,9,11):
    length = 30
else:
    print("Month out of range (1-12)")

print("Enter Date between 1-",length)
date = int(input())
if date>length:
    print("Invalid Date")
else:
    if date<length:
        date+=1
        print(f"Next Date:{date}-{y[month]}-{year}")
    else:
        date = 1
        if month == 12:
            month = 1
            year+=1
        else:
            month+=1
        print(f"Next Date:{date}-{y[month]}-{year}")
```

```python
# Harshad number
def harshad(num):
    temp = num
    sum = 0
    # Extract digits using math until the number becomes 0
    while num != 0:
        d = num % 10          # Gets the last digit (e.g., 19 % 10 = 9)
        sum += d        # adds it to the running total
        num = num // 10 
    
    if temp % sum == 0:
        print(temp,"is Harshad number")
    else:
        print(temp,"is not an Harshad number")

num = int(input("Enter Number :"))




harshad(num)
```

```python
import math

# Take input from the user
num = int(input("Enter a number to check: "))

# A pronic number is a product of two consecutive integers: n * (n + 1) = num
# This can be rewritten as: n^2 + n - num = 0
# Using the quadratic formula, we check if (1 + 4 * num) is a perfect square.
if num < 0:
    print(f"{num} is not a pronic number (pronic numbers are non-negative).")
else:
    discriminant = 1 + 4 * num
    root = math.isqrt(discriminant)

    # Check if it's a perfect square and the resulting 'n' is an integer
    if root * root == discriminant and (root - 1) % 2 == 0:
        n = (root - 1) // 2
        print(f"Yes, {num} is a pronic number!")
        print(f"It is the pronic of {n} (because {n} * {n + 1} = {num}).")
    else:
        print(f"No, {num} is not a pronic number.")
```
