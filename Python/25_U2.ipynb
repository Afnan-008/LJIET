```python
# Armstrong Number Checker
n = int(input("Enter Number: "))
temp = n

# 1. Count the number of digits
length = 0
while temp != 0:
    temp = temp // 10
    length += 1

# Reset temp back to the original number for calculation
temp = n
arm = 0

# 2. Calculate the Armstrong sum
while temp != 0:
    digit = temp % 10
    arm = arm + (digit ** length)
    temp = temp // 10

# 3. Check and print the result
if arm == n:
    print("Armstrong number")
else:
    print("Not an Armstrong number")
```

```python
#Prime Number Checker
n = int(input("Enter Number: "))

if n <= 1:
    is_prime = False
else:
    is_prime = True
    i = 2
    # i * i <= n is the exact same as checking up to the square root of n
    while i * i <= n:
        if n % i == 0:
            is_prime = False
            break
        i += 1

if is_prime:
    print("Prime")
else:
    print("Not Prime")
```

```python
# Get the upper limit from the user
limit = int(input("Enter the upper limit: "))

print(f"Prime numbers up to {limit}:")

# Loop through every number from 2 up to the user's limit
for n in range(2, limit + 1):
    is_prime = True
    i = 2
    
    # Check if the current number 'n' is prime
    while i * i <= n:
        if n % i == 0:
            is_prime = False
            break
        i += 1
        
    # If the number is prime, print it
    if is_prime:
        print(n,end=" ")
```

```python
# Happy number wothout str 
# Take input from the user
num = int(input("Enter a number to check: "))

# Keep a copy of the original number for the final print statement
original_num = num

# Track numbers we have already seen to catch endless loops
seen = set()

# Loop until the number becomes 1 or starts repeating
while num != 1 and num not in seen:
    seen.add(num)
    
    digit_sum = 0
    # Extract digits using math until the number becomes 0
    while num > 0:
        digit = num % 10          # Gets the last digit (e.g., 19 % 10 = 9)
        digit_sum += digit ** 2   # Squares and adds it to the running total
        num = num // 10           # Removes the last digit (e.g., 19 // 10 = 1)
        
    # Update num with the new sum for the next round
    num = digit_sum

# Check the final result
if num == 1:
    print(f"{original_num} is a Happy Number!")
else:
    print(f"{original_num} is NOT a Happy Number.")
```

```python
# Disarium Number Checker
n = int(input("Enter Number: "))
temp = n


# 1. Count the number of digits
length = 0
while temp != 0:
    temp = temp // 10
    length += 1

# Reset temp back to the original number for calculation
temp = n
dis = 0


# 2. Calculate the Disarium sum
while temp != 0:
    for i in range(length,0,-1):
        digit = temp % 10
        dis = dis + (digit ** i)
        temp = temp // 10
    

# 3. Check and print the result
if dis == n:
    print("Disarium number")
else:
    print("Not an Disarium number")
```

```python
# Get the upper limit from the user
limit = int(input("Enter the upper limit to check: "))

print(f"\nDisarium numbers up to {limit}:")

# Loop through each number from 1 to the limit
for n in range(1, limit + 1):
    
    # 1. Count the number of digits for the current number
    temp = n
    length = 0
    while temp != 0:
        temp = temp // 10
        length += 1

    # 2. Calculate the Disarium sum
    temp = n
    dis = 0
    while temp != 0:
        digit = temp % 10
        dis = dis + (digit ** length)
        length -= 1  # Decrease power as we move right to left
        temp = temp // 10

    # 3. Check and print if it matches
    if dis == n:
        print(n, end=" ")

```
