```python
#Aliacing and cloning of list
s = [1,2,3,4,5]
x=s
x[3]=100
print(s)
print(id(s))
print(id(x))
```

```python
#Aliacing and cloning of list
s = [1,2,3,4,5]
x=s[:]
x[3]=100
print(s)
print(x)
print(id(s))
print(id(x))
```

```python
#Aliacing and cloning of list
s = [1,2,3,4,5]
x=s.copy()
x[3]=100
print(s)
print(x)
print(id(s))
print(id(x))
```
