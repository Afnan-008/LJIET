```python
# Given a list l of size n. you need to count the number of special elements in the given list. an element is special if removal of that element makes 
# the list balanced. the list will be balanced if sum of even index is equal to sum of odd index element. also display the updated list after removal 
# of special elements.

l = [5,5,2,5,8]
special = []
count = 0
for i in range(0,len(l)):
    t = l[:]
    d = t.pop(i)
    sume = sum([t[i] for i in range(len(t)) if i % 2 == 0])
    sumo = sum([t[i] for i in range(len(t)) if i % 2 != 0])
    if sumo == sume:
        count+=1
        special.append(d)
print(count)
print(special)
```

```python
# WAP that reads the words from a paragraph and stores them as keys in a dictionary and count the frequency of it as its value.
paragraph = "dog the quick brown fox jumps over the lazy Dog and the dog remains asleep"
s = paragraph.split()
d = {}
for i in s:
    x = {i:s.count(i)}
    d.update(x)
print(d)
```

```python
paragraph = "dog the quick brown fox jumps over the lazy Dog and the dog remains asleep"
s = paragraph.split()
d={}
for i in s:
    element = []
    for j in s:
        if j[0] == i[0]:
            element.append(j)
    x = {i[0]:element}
    d.update(x)
print(d)           
```

```python
paragraph = "dog the quick brown fox jumps over the lazy Dog and the dog remains asleep"
words = paragraph.split()
d={}
for word in words:
    key = word[0]
    if key in d:
        d[key].append(word)
    else:
        d[key] = [word]
print(d)
    
```

```python
# WAP to replace the words in a given string from the look-up dictionary
s = "Don't wait for your feelings to change to take action Take action. asn your feelings will change"
d = {"action" : "lazy", "feelings" : "rude"}
new = []
for i in s.split():
    if i in d.keys():
        new.append(d[i])
    else:
        new.append(i)
x = " ".join(new)
print(s)
print(x)
```

### mathemetical operations

```python
# union() symbol '|'
s1 = {1,2,3}
s2 = {4,5,6,1}
s3 = {7,8,9}
print(s1.union(s2))
print(s2.union(s1))
print(s1.union(s1))
print(s1.union(s2,s3))
print()
print(s1|s2)
print(s1|s2|s3)
```

```python
# intersection symbol'&'
s1 = {1,2,3,4}
s2 = {4,5,6,1}
s3 = {7,8,9,5}

print(s1.intersection(s2))
print(s1&s2)
```

```python
# difference symbol '-' Returns unique elements.
s1 = {1,2,3,4}
s2 = {4,5,6,1,7}
print(s1.difference(s2))
print(s1-s2)
print(s2.difference(s1))
```

```python
# symmetric_difference() symbol '^'
s1 = {1,2,3,4}
s2 = {4,5,6,1,7}
print(s1.symmetric_difference(s2))
print(s1^s2)
```

```python
# issubset()
x = {'a',"b","c"}
y = {"e","f","a","b","c"}
print(x.issubset(y))
print(y.issubset(x))
```

```python
# issuperset()
x = {'a',"b","c"}
y = {"e","f","a","b","c"}
print(x.issuperset(y))
print(y.issuperset(x))
```
